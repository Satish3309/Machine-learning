```python
import pickle 
import numpy as np
import joblib
```

```python
# load saved standardscaler for input pre-processing
sc=joblib.load('scaler.pkl')
```

```python
preinput=sc.transform([[30,30,2,1,1]])
```

```python
loaded_model=pickle.load(open("finalized_model_svr.sav",'rb'))
result=loaded_model.predict(preinput)
```

```python
#load saved standardscaler for output reverse pre-processing 
s_cy=joblib.load('scaler_scy.pkl')
```

```python
preoutput=s_cy.inverse_transform([result])
```
